```python
#auxiliary functions
def atoi(text):
    return int(text) if text.isdigit() else text
def natural_keys(text):
    '''
    alist.sort(key=natural_keys) sorts in human order
    http://nedbatchelder.com/blog/200712/human_sorting.html
    (See Toothy's implementation in the comments)
    '''
    return [ atoi(c) for c in re.split(r'(\d+)', text) ]
```

```python
#folder containing original images and folder to save the new images
pathresults = "Dataset\results"; #path results
pathgt = "Dataset\labels"; #path labels

#list for metrics
dice_pred = []
ac_pred = []
pr_pred = []
re_pred = []

dice_pred_pos = []
ac_pred_pos = []
pr_pred_pos = []
re_pred_pos = []

#create the list of nii images in the directory
allfiles = os.listdir(os.path.join(pathresults))
allfiles.sort(key=natural_keys)
newlist = []
for names in allfiles:
    if names.endswith(".png"):
        newlist.append(names)

#process each image
for i in range(len(newlist)):
    predicted = imread(pathresults + "\\" + newlist[i])
    predicted = np.flip(np.rot90(predicted,1),0) # correct orientation
    gt = imread(pathgt + "\\" + newlist[i])

    # metrics without pos_processing
    dice, ac, pr, re = calculate_metrics(predicted,gt)
    dice_pred.append(dice)
    ac_pred.append(ac)
    pr_pred.append(pr)
    re_pred.append(re)

    # pos processing (e.g opening and labeling)
    predicted_pos_proc = pos_process(predicted)

    # metrics without pos_processing
    dice, ac, pr, re = calculate_metrics(predicted_pos_proc,gt)
    dice_pred_pos.append(dice)
    ac_pred_pos.append(ac)
    pr_pred_pos.append(pr)
    re_pred_pos.append(re)
```
