### I. Funções de Apresentação


#### 1. Função `mostrar_imagens()`

Função auxiliar para visualização de imagens.

- apresenta até 4 imagens lado a lado;
- permite adicionar títulos;
- aplica automaticamente colormap para imagens grayscale;
- reduz repetição de código no notebook.

```python
# ==========================================================
# 1. APRESENTAÇÃO DE IMAGENS (MODULAR)
# ==========================================================

def mostrar_imagens(imagens, titulos=None, mostrar_eixos=False, cmap="gray"):
    """
    Visualiza até 4 imagens alinhadas horizontalmente utilizando Matplotlib.

    Parâmetros:
        imagens (list): Lista contendo entre 1 e 4 arrays NumPy.
        titulos (list, opcional): Lista de títulos para cada imagem.
        mostrar_eixos (bool): Se True, apresenta os eixos; se False, oculta-os.
        cmap (str): Colormap a aplicar em imagens grayscale.
    """

    # Verifica se 'imagens' é uma lista
    if not isinstance(imagens, list):
        raise ValueError("O parâmetro 'imagens' deve ser uma lista.")

    # Permitir entre 1 e 4 imagens
    if len(imagens) < 1 or len(imagens) > 4:
        raise ValueError("A função aceita entre 1 e 4 imagens.")

    # Guarda o número de imagens a apresentar
    n = len(imagens)

    # Cria a figura com largura proporcional ao número de imagens
    # (5 unidades por imagem, altura fixa 5)
    plt.figure(figsize=(5 * n, 5))

    # Percorre cada imagem da lista
    for i in range(n):

        # Cria um subplot numa grelha de 1 linha e n colunas
        # i+1 indica a posição atual (index começa em 1 no subplot)
        plt.subplot(1, n, i + 1)

        img = imagens[i]

        # Se for imagem grayscale (2D), aplica cmap
        if img.ndim == 2:
            plt.imshow(img, cmap=cmap)
        else:
            # Se for RGB ou RGBA, não aplica cmap
            plt.imshow(img)

        # Se existirem títulos e houver título correspondente ao índice atual
        if titulos is not None and i < len(titulos):
            plt.title(titulos[i])

        # Se não for para mostrar eixos, remove-os
        if not mostrar_eixos:
            plt.axis("off")

    # Ajusta automaticamente espaçamentos internos da figura
    plt.tight_layout()

    # Renderiza a figura no output
    plt.show()
```

#### 2. Função `mostrar_tabela()`

Função auxiliar para apresentação de tabelas.

- apresenta dados estruturados em formato tabular;
- permite definir nomes das colunas;
- permite adicionar um título à tabela;
- ajusta automaticamente o tamanho da figura;
- reduz repetição de código na criação de tabelas com Matplotlib.

```python
# ==========================================================
# 2. APRESENTAÇÃO DE TABELAS (MODULAR)
# ==========================================================

def mostrar_tabela(dados, colunas=None, titulo=None, tamanho=(8, 3), mostrar=True):
    """
    Cria e apresenta uma tabela utilizando Matplotlib.

    Parâmetros:
        dados (list): Lista de listas contendo os dados organizados por linhas.
        colunas (list, opcional): Lista com os nomes das colunas.
        titulo (str, opcional): Título apresentado acima da tabela.
        tamanho (tuple): Dimensão da figura no formato (largura, altura).
        mostrar (bool): Se True, renderiza a tabela no output.
    """

    # Import interno (garante que a função funciona mesmo fora do contexto global)
    import matplotlib.pyplot as plt

    # Cria figura e eixo com o tamanho especificado (largura, altura)
    fig, ax = plt.subplots(figsize=tamanho)

    # Remove os eixos visuais (escala e grelha),
    # pois apenas queremos mostrar a tabela
    ax.axis("off")

    # Cria objeto tabela associado ao eixo atual
    tabela = ax.table(
        cellText=dados,     # Dados organizados por linhas
        colLabels=colunas,  # Cabeçalhos das colunas (se existirem)
        loc="center"        # Posiciona a tabela no centro da figura
    )

    # Desativa ajuste automático do tamanho da fonte
    tabela.auto_set_font_size(False)

    # Define tamanho fixo da fonte
    tabela.set_fontsize(10)

    # Ajusta escala das células (largura, altura)
    tabela.scale(1, 1.5)

    # Se existir título, adiciona-o à figura
    if titulo:
        plt.title(titulo)

    # Renderiza a figura apenas se o parâmetro 'mostrar' for True
    if mostrar:
        plt.show()

    # Devolve a figura e o objeto tabela
    # (permite personalização adicional fora da função)
    return fig, tabela
```

### II. Funções de Processamento

#### 3. Função `manipulacao_img()`

Função auxiliar para ajuste de brilho e contraste na imagem.

- aplica transformação linear \( g(i,j) = \alpha f(i,j) + b \);
- `alpha` controla o contraste da imagem;
- `b` controla o brilho da imagem;
- aplica a transformação apenas na região de interesse (`mask == 255`);
- mantém o fundo da imagem inalterado;
- garante que os valores permanecem no intervalo válido `[0,255]`.

```python
# ==========================================================
# 3. AJUSTE DE BRILHO E CONTRASTE (MANUAL)
# ==========================================================

def manipulacao_img(img, mask, alpha=1.0, b=0):
    """
    Aplica transformação linear na ROI segundo:

        g(i,j) = alpha * f(i,j) + b

    alpha → controla contraste
    b     → controla brilho

    A alteração é aplicada apenas na ROI (mask == 255).
    O fundo permanece inalterado.
    """

    # Criar cópia da imagem original
    nova_img = img.copy()

    # Obter dimensões
    altura, largura = img.shape

    # Percorrer todos os píxeis
    for i in range(altura):
        for j in range(largura):

            # Aplicar apenas na ROI
            if mask[i, j] == 255:

                # Valor original
                f_ij = int(img[i, j])

                # Aplicar transformação linear
                g_ij = alpha * f_ij + b

                # Garantir intervalo válido [0,255]
                if g_ij > 255:
                    g_ij = 255
                elif g_ij < 0:
                    g_ij = 0

                # Atualizar pixel
                nova_img[i, j] = int(g_ij)

    return nova_img
```

#### 4. Função `extrair_roi()`

Função auxiliar para extração da região de interesse (ROI) da imagem.

- identifica os píxeis da imagem onde `mask == valor`;
- extrai os valores de intensidade da ROI para um vetor (`roi_pixels`);
- cria uma imagem contendo apenas a região de interesse (`roi_img`);
- mantém o restante da imagem com valor zero;
- facilita análises posteriores como cálculo de histogramas da ROI.

```python
# ==========================================================
# 4. EXTRAIR REGIÃO DE INTERESSE (GENÉRICA)
# ==========================================================

def extrair_roi(img, mask, valor=255):
    """
    Extrai os pixels da imagem onde a máscara é igual ao valor especificado.

    Parâmetros:
        img (np.ndarray): Imagem original.
        mask (np.ndarray): Máscara associada.
        valor (int): Valor da máscara que define a ROI.

    Retorna:
        roi_pixels : lista com intensidades da ROI
        roi_img    : imagem contendo apenas a ROI
    """

    # Obter dimensões da imagem
    altura, largura = img.shape

    # Lista para guardar os píxeis da ROI
    roi_pixels = []

    # Criar imagem vazia com mesma dimensão da original
    roi_img = np.zeros((altura, largura), dtype=img.dtype)

    # Percorrer todos os píxeis da imagem
    for i in range(altura):
        for j in range(largura):

            # Verificar se o pixel pertence à ROI
            if mask[i, j] == valor:

                # Guardar intensidade do pixel
                roi_pixels.append(img[i, j])

                # Copiar pixel para a imagem ROI
                roi_img[i, j] = img[i, j]

    # Converter lista para array NumPy
    roi_pixels = np.array(roi_pixels)

    return roi_pixels, roi_img
```

#### 5. Função `calcular_histograma_manual_roi()`

Função auxiliar para cálculo manual do histograma da região de interesse (ROI).

- percorre a imagem pixel a pixel;
- considera apenas os píxeis pertencentes à ROI (`mask == 255`);
- conta a frequência de cada intensidade no intervalo `[0,255]`;
- devolve o histograma (`hist`) e os limites dos bins (`bin_edges`);
- permite analisar a distribuição de intensidades na região de interesse.

```python
# ==========================================================
# 5. HISTOGRAMA MANUAL DA ROI (IMG + MASK)
# ==========================================================

def calcular_histograma_manual_roi(img, mask):
    """
    Calcula manualmente o histograma da Região de Interesse (ROI)
    percorrendo a imagem pixel a pixel.

    A máscara define a região espacial válida.
    A imagem original fornece os valores de intensidade.
    """

    # Criar vetor com 256 posições (0–255), inicializado a zero
    hist = np.zeros(256, dtype=int)

    # Obter dimensões da imagem
    altura, largura = img.shape

    # Percorrer todos os píxeis da imagem
    for i in range(altura):
        for j in range(largura):

            # Considerar apenas píxeis pertencentes à ROI
            if mask[i, j] == 255:

                # Obter intensidade real da imagem original
                intensidade = int(img[i, j])

                # Incrementar frequência correspondente
                hist[intensidade] += 1

    # Criar vetor de limites compatível com np.histogram
    bin_edges = np.arange(257)

    # Devolver histograma e limites
    return hist, bin_edges
```

#### 1. Função `criar_uploaders()`

Função auxiliar para criação de widgets de upload de imagens.

- cria dois componentes `FileUpload` para seleção da imagem e da máscara;
- apresenta os widgets diretamente no notebook;
- permite ao utilizador selecionar ficheiros locais para processamento;
- devolve os widgets para acesso posterior aos ficheiros carregados.

```python
# ==========================================================
# 1. CRIAR UPLOADERS
# ==========================================================

def criar_uploaders():
    """
    Cria dois widgets FileUpload para seleção da imagem
    principal e da respetiva máscara.
    """

    # ---------------------------------------------------------
    # BOTÃO 1 – Upload da imagem principal
    # ---------------------------------------------------------

    # Cria um widget interativo que permite ao utilizador
    # selecionar um ficheiro do computador.
    upload_img = widgets.FileUpload(
        accept='image/*',      # aceita apenas ficheiros de imagem
        multiple=False,        # permite selecionar apenas um ficheiro
        description='Imagem'   # texto apresentado no botão
    )

    # ---------------------------------------------------------
    # BOTÃO 2 – Upload da máscara
    # ---------------------------------------------------------

    # Widget semelhante ao anterior, mas destinado à máscara.
    upload_mask = widgets.FileUpload(
        accept='image/*',
        multiple=False,
        description='Máscara'
    )

    # ---------------------------------------------------------
    # APRESENTAR OS WIDGETS NO NOTEBOOK
    # ---------------------------------------------------------

    # A função display() torna os widgets visíveis
    # na célula do Jupyter Notebook.
    display(upload_img)
    display(upload_mask)

    # ---------------------------------------------------------
    # DEVOLVER OS WIDGETS CRIADOS
    # ---------------------------------------------------------

    # Isto permite aceder posteriormente aos ficheiros
    # selecionados pelo utilizador.
    return upload_img, upload_mask
```

#### 3. Função `carregar_imagem()`

Função auxiliar para conversão de ficheiros carregados através
dos widgets `FileUpload` em arrays NumPy utilizáveis.

- verifica se existe um ficheiro selecionado;
- extrai o conteúdo binário do widget;
- trata diferentes formatos internos (`dict` ou `tuple`);
- converte os dados binários numa imagem utilizando `matplotlib`.

```python
# ==========================================================
# 3. CARREGAR IMAGENS
# ==========================================================

def carregar_imagem(upload_widget):
    """
    Converte o ficheiro selecionado num widget FileUpload
    num array NumPy utilizável.
    """

    # ---------------------------------------------------------
    # VERIFICAR SE EXISTE FICHEIRO SELECIONADO
    # ---------------------------------------------------------

    # Se o utilizador ainda não tiver escolhido um ficheiro,
    # o atributo value estará vazio.
    if not upload_widget.value:
        return None

    # Guarda o conteúdo interno do widget
    valor = upload_widget.value


    # ---------------------------------------------------------
    # EXTRAIR CONTEÚDO DO FICHEIRO
    # ---------------------------------------------------------

    # Dependendo da versão do Jupyter,
    # o conteúdo pode vir em formatos diferentes.

    # Caso 1 – estrutura em dicionário
    if isinstance(valor, dict):

        ficheiro = list(valor.values())[0]
        conteudo = ficheiro['content']

    # Caso 2 – estrutura em tupla
    elif isinstance(valor, tuple):

        ficheiro = valor[0]
        conteudo = ficheiro['content']

    # Caso inesperado
    else:
        return None


    # ---------------------------------------------------------
    # CONVERTER BYTES EM IMAGEM
    # ---------------------------------------------------------

    # BytesIO cria um "ficheiro virtual" em memória
    # que pode ser lido pelo matplotlib.
    return mpimg.imread(io.BytesIO(conteudo))
```

### II. Analisar e Converter Metadados

> Esta secção assegura a normalização e padronização estrutural das imagens carregadas, garantindo que se encontram num formato consistente e adequado para processamento posterior.
>
> O objetivo principal é eliminar variações estruturais (formato de cor e tipo de dados) que possam comprometer etapas seguintes da análise.


#### 1. Função `analisar_metadados()`

Função auxiliar para análise estrutural da imagem.

- identifica dimensão (`shape`) da imagem;
- identifica tipo de dados (`dtype`);
- determina o número de canais da imagem;
- facilita a verificação do estado da imagem antes e após processamento.

```python
# ==========================================================
# 1. ANALISAR METADADOS ORIGINAIS
# ==========================================================

def analisar_metadados(img):
    """
    Analisa dimensão, tipo de dados e número de canais da imagem.
    """

    # Se a imagem tiver apenas 2 dimensões → grayscale → 1 canal
    # Caso contrário assume imagem com múltiplos canais
    canais = 1 if img.ndim == 2 else img.shape[2]

    # Devolve metadados estruturados
    return {
        "Shape": img.shape,
        "Dtype": img.dtype,
        "Canais": canais
    }


# ---------------------------------------------------------
# EXECUÇÃO
# ---------------------------------------------------------

# Analisar metadados da imagem e da máscara
antes_img = analisar_metadados(img)
antes_mask = analisar_metadados(img_mask)

print("=== Metadados Iniciais ===")
print("Imagem:", antes_img)
print("Máscara:", antes_mask)
```

#### 2. Função `converter_para_grayscale()`

Função auxiliar para conversão de imagens para escala de cinzentos.

- verifica se a imagem já se encontra em grayscale;
- converte imagens RGB/RGBA utilizando ponderação de luminância;
- preserva altura e largura da imagem;
- prepara a imagem para processamento de intensidades.

```python
# ==========================================================
# 2. CONVERTER PARA GRAYSCALE
# ==========================================================

def converter_para_grayscale(img):
    """
    Converte imagem RGB/RGBA para grayscale
    preservando as dimensões espaciais.
    """

    # Se já for grayscale, devolve diretamente
    if img.ndim == 2:
        return img

    # Extrair canais RGB
    r = img[:, :, 0]
    g = img[:, :, 1]
    b = img[:, :, 2]

    # Aplicar fórmula de luminância ponderada
    gray = 0.299*r + 0.587*g + 0.114*b

    return gray


# ---------------------------------------------------------
# EXECUÇÃO
# ---------------------------------------------------------

# Converter imagem e máscara para grayscale
img_gray = converter_para_grayscale(img)
mask_gray = converter_para_grayscale(img_mask)

print("Conversão para grayscale concluída.")
```

#### 3. Função `garantir_uint8()`

Função auxiliar para normalização do tipo de dados da imagem.

- verifica se a imagem já se encontra no formato `uint8`;
- converte imagens normalizadas (0–1) para intervalo 0–255;
- limita valores ao intervalo válido de intensidades;
- garante compatibilidade com operações de processamento de imagem.

```python
# ==========================================================
# 3. GARANTIR FORMATO UINT8 (PROCESSO MANUAL)
# ==========================================================

def garantir_uint8_manual(img):
    """
    Converte imagem normalizada (0–1) para uint8 (0–255).
    """

    # Se já estiver em uint8, devolve diretamente
    if img.dtype == np.uint8:
        return img

    # Escalar intensidades de [0,1] para [0,255]
    img_escalada = img * 255

    # Arredondar valores
    img_arredondada = np.round(img_escalada)

    # Garantir intervalo válido
    img_limitada = np.clip(img_arredondada, 0, 255)

    # Converter para uint8
    img_uint8 = img_limitada.astype(np.uint8)

    return img_uint8


# ---------------------------------------------------------
# EXECUÇÃO
# ---------------------------------------------------------

# Converter imagem e máscara
img_uint8 = garantir_uint8_manual(img_gray)
mask_uint8 = garantir_uint8_manual(mask_gray)

print("Conversão manual para uint8 concluída.")
```

#### 4. Comparação de Metadados

Comparação da estrutura da imagem antes e após o pré-processamento.

- reanalisa dimensão, tipo de dados e canais;
- organiza os resultados numa tabela;
- permite validar a transformação para grayscale e `uint8`.

```python
# ==========================================================
# 4. COMPARAÇÃO ANTES / DEPOIS
# ==========================================================

# Reanalisar metadados após processamento
depois_img = analisar_metadados(img_uint8)
depois_mask = analisar_metadados(mask_uint8)


# ---------------------------------------------------------
# PREPARAR DADOS PARA TABELA
# ---------------------------------------------------------

dados = [
    ["Imagem - Antes",  antes_img["Shape"],  antes_img["Dtype"],  antes_img["Canais"]],
    ["Imagem - Depois", depois_img["Shape"], depois_img["Dtype"], depois_img["Canais"]],
    ["Máscara - Antes",  antes_mask["Shape"],  antes_mask["Dtype"],  antes_mask["Canais"]],
    ["Máscara - Depois", depois_mask["Shape"], depois_mask["Dtype"], depois_mask["Canais"]],
]

colunas = ["Tipo", "Shape", "Dtype", "Canais"]


# ---------------------------------------------------------
# MOSTRAR TABELA
# ---------------------------------------------------------

mostrar_tabela(
    dados=dados,
    colunas=colunas,
    titulo="Comparação de Metadados - Antes vs Depois",
    tamanho=(10,4)
)


# ---------------------------------------------------------
# ATUALIZAR VARIÁVEIS PRINCIPAIS
# ---------------------------------------------------------

# Substitui imagens originais pelas versões processadas
img = img_uint8
mask = mask_uint8
```

### III. Verificar Compatibilidade Estrutural

> Esta secção valida a coerência estrutural entre a imagem a analisar e a respetiva máscara selecionada, assegurando que ambas são compatíveis para processamento conjunto.
>
> O objetivo é detetar inconsistências dimensionais, estruturais ou semânticas que possam comprometer a fiabilidade da segmentação.


#### 1. Função `verificar_dimensao()`

Função auxiliar para validação dimensional entre imagem e máscara.

- compara o `shape` da imagem e da máscara;
- confirma se ambas possuem a mesma dimensão espacial;
- evita inconsistências nas operações baseadas em máscara.

```python
# ==========================================================
# 1. VERIFICAR CORRESPONDÊNCIA DIMENSIONAL
# ==========================================================

def verificar_dimensao(img, mask):
    """
    Verifica se imagem e máscara possuem exatamente o mesmo shape.
    """

    # Comparar dimensões diretamente
    return img.shape == mask.shape


# ---------------------------------------------------------
# EXECUÇÃO
# ---------------------------------------------------------

dimensao_ok = verificar_dimensao(img, mask)

if not dimensao_ok:
    print("Dimensão incompatível.")
    print("\n** Por favor selecione outra mascara! **\n")
else:
    print("Dimensão compatível.")
```

#### 2. Função `validar_mascara_binaria()`

Função auxiliar para validação da binariedade da máscara.

- verifica se a máscara contém apenas valores `0` e `255`;
- calcula a percentagem de píxeis não binários;
- permite uma tolerância configurável;
- aplica correção automática por threshold quando necessário.

```python
# ==========================================================
# 2. VALIDAR BINARIEDADE DA MÁSCARA
# ==========================================================

def validar_mascara_binaria(mask, tolerancia=0.02):
    """
    Verifica se a máscara é binária (0 e 255).
    Permite pequena tolerância de valores não binários.
    """

    # Número total de píxeis
    total = mask.size

    # Contar píxeis diferentes de 0 e 255
    nao_binarios = np.sum((mask != 0) & (mask != 255))

    # Percentagem de valores inválidos
    percentagem = nao_binarios / total

    # Se ultrapassar tolerância → máscara inválida
    if percentagem > tolerancia:
        return False, mask

    # Corrigir automaticamente via threshold
    mask_corrigida = np.zeros_like(mask, dtype=np.uint8)

    try:
        for i in range(mask.shape[0]):
            for j in range(mask.shape[1]):
                if mask[i, j] < 127:
                    mask_corrigida[i, j] = 0
                else:
                    mask_corrigida[i, j] = 255
        return True, mask_corrigida
        
    except:
        return False, mask


# ---------------------------------------------------------
# EXECUÇÃO
# ---------------------------------------------------------

mascara_valida, mask_corrigida = validar_mascara_binaria(mask)

if not mascara_valida:
    print("Máscara contém demasiados valores não binários")
    print("\n** Por favor selecione outra mascara! **\n")
else:
    print("Máscara válida (com tolerância).")
    mask = mask_corrigida
```

#### 4. Função `verificar_fundo()`

Função auxiliar para validação da coerência entre máscara e imagem.

- identifica regiões de fundo (`mask == 0`);
- verifica se essas regiões são suficientemente escuras na imagem;
- calcula a percentagem de incoerências;
- valida a correspondência com base numa tolerância definida.

```python
# ==========================================================
# 3. VERIFICAR COERÊNCIA DO FUNDO
# ==========================================================

def verificar_fundo(img, mask, limiar_preto=30, tolerancia=0.02):
    """
    Verifica se regiões de fundo da máscara correspondem
    a zonas escuras na imagem.
    """

    # Identificar região de fundo
    fundo_mask = (mask == 0)

    # Número total de píxeis de fundo
    total_fundo = np.sum(fundo_mask)

    # Se não existir fundo → não é possível validar
    if total_fundo == 0:
        return False
        
    try:
        # Contar píxeis de fundo demasiado claros
        incoerencias = np.sum((img >= limiar_preto) & fundo_mask)
    
        # Percentagem de incoerência
        percentagem = incoerencias / total_fundo
    
        return percentagem <= tolerancia
        
    except:
        return False


# ---------------------------------------------------------
# EXECUÇÃO
# ---------------------------------------------------------

fundo_ok = verificar_fundo(img, mask)

if not fundo_ok:
    print("Incoerência significativa no fundo da máscara.")
    print("\n** Por favor selecione outra mascara! **\n")
else:
    print("Máscara estruturalmente compatível (com tolerância).")
```

## Question 5
> Invert the image. The black background should not be changed as it does not belong to the region of interest.

```python
# ==========================================================
# 1. INVERSÃO NA REGIÃO DE INTERESSE (ROI)
# ==========================================================

def inverter_roi(img, mask):
    """
    Inverte intensidades apenas na ROI:

        g(i,j) = 255 - f(i,j)

    Apenas onde mask == 255.
    Fundo permanece inalterado.
    """

    nova_img = img.copy()

    nova_img[mask == 255] = 255 - nova_img[mask == 255]

    return nova_img
```

```python
# ==========================================================
# 2. EXECUÇÃO – INVERSÃO DA IMAGEM
# ==========================================================

roi_pixels, roi_img = extrair_roi(img, mask)

if roi_pixels.size == 0:

    print("ROI vazia. Não é possível inverter.")

else:

    img_invertida = inverter_roi(img, mask)
```

```python
# ==========================================================
# HISTOGRAMAS DAS IMAGENS
# ==========================================================

hist_original, bins = calcular_histograma_manual_roi(img, mask)
hist_invertida, _ = calcular_histograma_manual_roi(img_invertida, mask)

x = bins[:-1]
```

```python
# ==========================================================
# VISUALIZAÇÃO: IMAGEM | HISTOGRAMA
# ==========================================================

imagens = [img, img_invertida]
histogramas = [hist_original, hist_invertida]

titulos = [
    "Imagem Original",
    "Imagem Invertida (ROI)"
]

plt.figure(figsize=(12,6))

for i in range(len(imagens)):

    # IMAGEM
    plt.subplot(len(imagens), 2, i*2 + 1)
    plt.imshow(imagens[i], cmap="gray")
    plt.title(titulos[i])
    plt.axis("off")

    # HISTOGRAMA
    plt.subplot(len(imagens), 2, i*2 + 2)
    plt.bar(x, histogramas[i])
    plt.title(f"Histograma - {titulos[i]}")
    plt.xlabel("Intensidade")
    plt.ylabel("Frequência")
    plt.xlim(0,255)
    plt.grid(True)

plt.tight_layout()
plt.show()
```

- Desta forma, o histograma apresenta uma **simetria invertida relativamente ao centro da escala de intensidades**, mantendo-se as mesmas frequências, mas refletidas ao longo do intervalo de intensidade.

## Question 6
> Transform the image according to the histogram equalization technique. The black background should not be changed as it does not belong to the region of interest. Show the final image and its histogram, analyzing the differences to the original image.

```python
# ==========================================================
# HISTOGRAMA ORIGINAL DA ROI
# ==========================================================

hist_original, bins = calcular_histograma_manual_roi(img, mask)

x = bins[:-1] # excepto o ultimo para ser de 0 a 255
```

```python
# ==========================================================
# HISTOGRAMA CUMULATIVO
# ==========================================================

H = np.zeros_like(hist_original)

acumulado = 0

for i in range(len(hist_original)):
    acumulado += hist_original[i]
    H[i] = acumulado
```

```python
# ==========================================================
# TABELA DE TRANSFORMAÇÃO
# ==========================================================

G = 255                 # níveis de cinzento
S = roi_pixels.size     # número de píxeis da ROI

T = np.zeros(256)

for g in range(256):
    T[g] = round((G / S) * H[g])
```

```python
# ==========================================================
# APLICAR EQUALIZAÇÃO
# ==========================================================

img_equalizada = img.copy()

altura, largura = img.shape

for i in range(altura):
    for j in range(largura):

        if mask[i, j] == 255:

            intensidade = img[i, j]

            img_equalizada[i, j] = T[intensidade]
```

```python
# ==========================================================
# HISTOGRAMA DA IMAGEM EQUALIZADA
# ==========================================================

hist_eq, _ = calcular_histograma_manual_roi(img_equalizada, mask)
```

```python
# ==========================================================
# VISUALIZAÇÃO
# ==========================================================

imagens = [img, img_equalizada]
histogramas = [hist_original, hist_eq]

titulos = ["Imagem Original", "Imagem Equalizada"]

plt.figure(figsize=(12,8))

for i in range(len(imagens)):

    # imagem
    plt.subplot(len(imagens),2,i*2+1)
    plt.imshow(imagens[i], cmap="gray")
    plt.title(titulos[i])
    plt.axis("off")

    # histograma
    plt.subplot(len(imagens),2,i*2+2)
    plt.bar(x, histogramas[i])
    plt.title(f"Histograma - {titulos[i]}")
    plt.xlabel("Intensidade")
    plt.ylabel("Frequência")
    plt.xlim(0,255)
    plt.grid(True)

plt.tight_layout()
plt.show()
```

### Observações

#### Equalização do histograma nas imagens `breast_image` e `breast_mask`

- A equalização do histograma foi aplicada às imagens `breast_image` e `breast_mask`, considerando apenas os píxeis pertencentes à região de interesse (ROI). A transformação foi calculada usando G = 255 níveis de cinzento, S como o número total de píxeis da ROI e H(g) como o histograma acumulado.

- Na imagem original, o histograma apresenta uma forte concentração de frequências em determinadas intensidades, o que limita o contraste da imagem.

- Após a equalização, verifica-se que o histograma se distribui por uma gama maior de intensidades (0–255), reduzindo a concentração em intervalos específicos.

- Conclui-se que esta transformação aumenta o contraste da imagem, uma vez que redistribui as intensidades e melhora a separação entre diferentes tons, permitindo observar com maior clareza algumas estruturas presentes na ROI.
