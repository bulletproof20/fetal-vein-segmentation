**Question 5**: Use the labelling to differentiate the objects of the image. You must obtain one image per object.

```python
from skimage.measure import label, regionprops

img = plt.imread('structures.png')

label_img = label(img.astype(np.uint8))

plt.imshow(label_img)
plt.show()
```

```python
regions = regionprops(label_img)

count = 1
index_t = 0
index_r = 0
index_q = 0
index_c_p = 0
index_c_g = 0

array_circle = [];
array_circle_area = [];

for props in regions: #loop to print centroid of objects
      
    eccentricity = props.eccentricity
    area = props.area
    bbox_area = props.area_bbox
    axis_major_length = props.axis_major_length*2
    axis_minor_length = props.axis_minor_length*2

    if bbox_area == area:
        if abs(axis_major_length - axis_minor_length) < 25:
                index_q = count
        else:
                index_r = count
    else:
        if eccentricity<0.5:
            array_circle.append(count)
            array_circle_area.append(area)
        else:
            index_t = count
                   
    count = count + 1

index_c_p = array_circle[np.argmin(array_circle_area)]
index_c_g = array_circle[np.argmax(array_circle_area)]
```

```python
t = label_img == index_t
r = label_img == index_r
q = label_img == index_q
c_p = label_img == index_c_p
c_g = label_img == index_c_g

#print das imagens de cada objeto
fig, (ax1,ax2,ax3,ax4,ax5)=matplotlib.pylab.subplots(1,5,figsize=(16,4))
gray()
ax1.imshow(t)
ax2.imshow(r)
ax3.imshow(q)
ax4.imshow(c_p)
ax5.imshow(c_g)
plt.show()
```
